This notebook demonstrates **Data Augmentation** techniques using `ImageDataGenerator` to improve model performance on the Corn Leaf Disease dataset.

## 1. Import Libraries

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
import os
from tensorflow.keras.preprocessing import image
```

## 2. Data Augmentation Configuration

```python
# Data Augmentation Generator
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)
```

## 3. Visualize Augmented Images

```python
# Load one image from Cercospora class
fnames = [os.path.join(train_Cercospora_dir, fname) 
          for fname in os.listdir(train_Cercospora_dir)]
img_path = fnames[3]   # You can change the index

img = image.load_img(img_path, target_size=(150, 150))
x = image.img_to_array(img)
x = x.reshape((1,) + x.shape)

# Generate and display augmented images
i = 0
for batch in datagen.flow(x, batch_size=1):
    plt.figure(i)
    imgplot = plt.imshow(image.array_to_img(batch[0]))
    i += 1
    if i % 4 == 0:   # Display 4 augmented images
        break
plt.show()
```
